```python
import re
from typing import Optional


class Time:
    """
    Clase que representa una hora en formato AM/PM o 24 horas.

    Atributos de clase:
        TIME_FORMATS: formatos permitidos ("AM", "PM", "24 HOURS").
        time_count = 0  # Cuenta el número de objetos Time creados

    Atributos:
        # Los atributos son privados y se exponen mediante property
        hours        # Guarda las horas
                     # (1 a 12 en AM/PM, 0 a 23 en 24 HOURS)
        minutes      # Guarda los minutos (0 a 59)
        seconds      # Guarda los segundos (0 a 59)
        time_format  # Guarda el formato: "AM", "PM" o "24 HOURS"
    """

    TIME_FORMATS: tuple[str, ...] = ("AM", "PM", "24 HOURS")
    time_count: int = 0

    # Método de instancia: cada objeto Time necesita su propio estado (self).
    def __init__(self) -> None:
        """
        Inicializa los atributos a 0.
        Es el único método que se le pasa parámetros como ejemplo
        """
        self.__hours: int = 0
        self.__minutes: int = 0
        self.__seconds: int = 0
        self.__time_format: str = "24 HOURS"   # 00:00:00 es una hora válida en 24 HOURS
        # Se incrementa el atributo de CLASE a través de Time: con self.time_count += 1
        # se crearía un atributo de instancia nuevo y el contador de clase no cambiaría.
        Time.time_count += 1

    # ---------------------------------------------------------------- properties
    # Propiedades (métodos de instancia con @property): exponen los atributos
    # privados y los setters validan el valor antes de guardarlo.
    @property
    def hours(self) -> int:
        """Horas (1 a 12 en AM/PM, 0 a 23 en 24 HOURS)."""
        return self.__hours

    @hours.setter
    def hours(self, value: int) -> None:
        if not self._is_valid_time(value, self.__minutes, self.__seconds):
            raise ValueError(f"Horas no válidas para el formato {self.__time_format}: {value}")
        self.__hours = value

    @property
    def minutes(self) -> int:
        """Minutos (0 a 59)."""
        return self.__minutes

    @minutes.setter
    def minutes(self, value: int) -> None:
        if not self._is_valid_time(self.__hours, value, self.__seconds):
            raise ValueError(f"Minutos no válidos: {value}")
        self.__minutes = value

    @property
    def seconds(self) -> int:
        """Segundos (0 a 59)."""
        return self.__seconds

    @seconds.setter
    def seconds(self, value: int) -> None:
        if not self._is_valid_time(self.__hours, self.__minutes, value):
            raise ValueError(f"Segundos no válidos: {value}")
        self.__seconds = value

    @property
    def time_format(self) -> str:
        """Formato de la hora: "AM", "PM" o "24 HOURS"."""
        return self.__time_format

    @time_format.setter
    def time_format(self, value: str) -> None:
        old_format = self.__time_format
        if not self.__assign_format(value):
            raise ValueError(f"Formato no válido: {value}")
        if not self._is_valid_time(self.__hours, self.__minutes, self.__seconds):
            self.__time_format = old_format      # se restaura: la hora actual no es válida en el nuevo formato
            raise ValueError(f"La hora actual no es válida en el formato {value}")

    # ------------------------------------------------------------------- métodos
    # Método de instancia (privado): modifica el atributo propio __time_format del objeto.
    def __assign_format(self, time_format: str) -> bool:
        """
        Comprueba que time_format tiene un valor correcto y lo
        asigna al atributo correspondiente.
        Lo convierte a mayúsculas para evitar problemas de
        capitalización.

        Args:
            time_format: cadena con el formato de hora
                         ("AM", "PM" o "24 HOURS").

        Devuelve:
            True si el formato es correcto, False en caso
            contrario.
        """
        if not isinstance(time_format, str):
            return False
        normalized = time_format.strip().upper()
        if not Time.is_valid_format(normalized):
            return False
        self.__time_format = normalized
        return True

    # Método de instancia (privado): su resultado depende del formato guardado en este objeto.
    def __is_24hour_format(self) -> bool:
        """
        Comprueba si el formato de la hora es "24 HOURS".

        Devuelve:
            True si formato es "24 HOURS", False en caso contrario.
        """
        return self.__time_format == "24 HOURS"

    # Método de instancia (protegido, un solo "_"): valida según el formato del objeto
    # (self) y está pensado para uso interno y de las subclases.
    def _is_valid_time(self, hours: int, minutes: int, seconds: int) -> bool:
        """
        Comprueba si la hora es correcta según el formato.

        Args:
            hours: horas a comprobar.
            minutes: minutos a comprobar.
            seconds: segundos a comprobar.

        Devuelve:
            True si la hora es correcta, False en caso contrario.
        """
        for value in (hours, minutes, seconds):
            if not isinstance(value, int) or isinstance(value, bool):   # bool es subclase de int
                return False
        if self.__is_24hour_format():
            hours_ok = 0 <= hours <= 23
        else:
            hours_ok = 1 <= hours <= 12
        return hours_ok and 0 <= minutes <= 59 and 0 <= seconds <= 59

    # Método de instancia: cambia el estado (hora y formato) del objeto sobre el que se llama.
    def set_time(self, hours: int, minutes: int, seconds: int, time_format: str) -> bool:
        """
        Asigna una hora.

        Args:
            hours: horas (1 a 12 en AM/PM, 0 a 23 en 24 HOURS).
            minutes: minutos (0 a 59).
            seconds: segundos (0 a 59).
            time_format: formato ("AM", "PM" o "24 HOURS").

        Devuelve:
            True si la hora se pudo asignar correctamente,
            False en caso contrario.
        """
        old_format = self.__time_format
        if not self.__assign_format(time_format):
            return False
        if not self._is_valid_time(hours, minutes, seconds):
            self.__time_format = old_format      # no se deja el objeto a medio modificar
            return False
        self.__hours = hours
        self.__minutes = minutes
        self.__seconds = seconds
        return True

    # Método de instancia: lee el estado de un objeto concreto.
    def get_time(self) -> tuple[int, int, int, str]:
        """
        Devuelve la hora actual del objeto.

        Devuelve:
            Una tupla (horas, minutos, segundos, formato).
        """
        return self.__hours, self.__minutes, self.__seconds, self.__time_format

    # Método de clase (@classmethod): es un constructor alternativo. No necesita un objeto
    # previo y recibe la clase (cls) para poder crear instancias con cls().
    @classmethod
    def from_string(cls, time_string: str) -> Optional["Time"]:
        """
        Crea una nueva instancia de Time a partir de una cadena.

        Args:
            time_string: cadena con la hora en el formato
                "HH:MM:SS FORMATO", donde FORMATO es AM, PM o
                24 HOURS.

        Devuelve:
            Una instancia de Time con la hora leída.
            Si la cadena no es válida, muestra un mensaje y devuelve None.
        """
        # Patrón que reconoce cadenas como "14:30:00 24 HOURS" o "02:45:30 PM"
        pattern = r"(\d{2}):(\d{2}):(\d{2})\s+(AM|PM|24 HOURS)"
        match = None
        if isinstance(time_string, str):
            match = re.fullmatch(pattern, time_string.strip(), re.IGNORECASE)
        if match is None:
            print('Cadena no válida. Use el formato "HH:MM:SS FORMATO" (AM, PM o 24 HOURS).')
            return None
        # Extraemos horas, minutos, segundos y formato de los grupos
        hours, minutes, seconds = (int(group) for group in match.group(1, 2, 3))
        time_format = match.group(4)
        instance = cls()   # (cuenta como objeto creado en time_count aunque luego se descarte)
        if not instance.set_time(hours, minutes, seconds, time_format):
            print(f"La hora '{time_string.strip()}' está fuera de rango para el formato {time_format.upper()}.")
            return None
        return instance

    # Método de clase (@classmethod): consulta un dato de la clase (TIME_FORMATS) sin necesitar
    # ningún objeto; con cls se respeta si una subclase redefine TIME_FORMATS.
    @classmethod
    def is_valid_format(cls, time_format: str) -> bool:
        """
        Comprueba si un formato de hora dado es
        válido.

        Args:
            time_format: cadena con el formato de hora a comprobar.

        Devuelve:
            True si el formato es válido (AM, PM o 24 HOURS),
            False en caso contrario.
        """
        return isinstance(time_format, str) and time_format.strip().upper() in cls.TIME_FORMATS

    # Método de clase (@classmethod): lee el contador de clase time_count, que es compartido
    # por todos los objetos y no pertenece a ninguno en concreto.
    @classmethod
    def get_time_count(cls) -> int:
        """
        Devuelve el número de objetos de la clase que se han
        creado.

        Devuelve:
            Un número entero de objetos Time creados.
        """
        return cls.time_count


def format_time(time_obj: Time) -> str:
    """Devuelve la hora de un objeto Time como cadena con formato "HH:MM:SS FORMATO"."""
    hours, minutes, seconds, time_format = time_obj.get_time()
    return f"{hours:02d}:{minutes:02d}:{seconds:02d} {time_format}"


def read_int(prompt: str, minimum: int, maximum: int) -> int:
    """Pide un entero por teclado hasta que sea válido y esté entre minimum y maximum (incluidos)."""
    while True:
        try:
            value = int(input(prompt))
        except ValueError:
            print("Entrada no válida. Introduzca un número entero.")
            continue
        if minimum <= value <= maximum:
            return value
        print(f"Valor fuera de rango. Debe estar entre {minimum} y {maximum}.")


def read_format() -> str:
    """Pide un formato de hora hasta que sea válido (se acepta en cualquier caso) y lo devuelve en mayúsculas."""
    while True:
        text = input('Formato ("AM", "PM" o "24 HOURS"): ')
        if Time.is_valid_format(text):
            return text.strip().upper()
        print('Formato no válido. Debe ser "AM", "PM" o "24 HOURS".')


def enter_new_time(time_obj: Time) -> None:
    """Pide una hora completa al usuario, validando cada dato, y la asigna a time_obj."""
    time_format = read_format()
    min_hours, max_hours = (0, 23) if time_format == "24 HOURS" else (1, 12)
    hours = read_int(f"Horas ({min_hours} a {max_hours}): ", min_hours, max_hours)
    minutes = read_int("Minutos (0 a 59): ", 0, 59)
    seconds = read_int("Segundos (0 a 59): ", 0, 59)
    if time_obj.set_time(hours, minutes, seconds, time_format):
        print(f"Hora asignada: {format_time(time_obj)}")
    else:
        print("No se pudo asignar la hora.")


MENU = """
--- MENÚ ---
1. Introducir una nueva hora
2. Visualizar hora
3. Crear una hora a partir de una cadena ("HH:MM:SS FORMATO")
4. Terminar"""


def main() -> None:
    """Ejecuta el menú del programa hasta que el usuario elija terminar."""
    current = Time()
    while True:
        print(MENU)
        option = input("Opción: ").strip()
        if option == "1":
            enter_new_time(current)
        elif option == "2":
            print(f"Hora actual: {format_time(current)}")
            print(f"Objetos Time creados hasta ahora: {Time.get_time_count()}")
        elif option == "3":
            new_time = Time.from_string(input('Hora ("HH:MM:SS FORMATO"): '))
            if new_time is not None:
                current = new_time
                print(f"Hora creada: {format_time(current)}")
        elif option == "4":
            print("Hasta pronto.")
            break
        else:
            print("Opción no válida. Elija 1, 2, 3 o 4.")
```
