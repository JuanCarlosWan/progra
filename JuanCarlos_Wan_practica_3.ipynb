````python
# utils.py: funciones de lectura de datos y menú
import math


def leer_int(mensaje: str = "Introduce un número entero: ") -> int:
    """
    Función auxiliar que lee un número entero del teclado.
    Si se introduce un valor no válido, se solicita de nuevo.

    Parámetros:
        mensaje: el mensaje que se muestra al usuario solicitando la entrada (str).
                 Por defecto "Introduce un número entero: "

    Devuelve:
        int: El valor entero introducido por el usuario.
    """
    while True:
        try:
            return int(input(mensaje))
        except ValueError:
            print("Entrada no válida. Debe ser un número entero.")


def leer_float(mensaje: str = "Introduce un número decimal: ") -> float:
    """
    Función auxiliar que solicita al usuario un número decimal.
    Si se introduce un valor no válido, se solicita de nuevo.

    Parámetros:
        mensaje: el mensaje que se muestra al usuario solicitando la entrada (str).
                 Por defecto "Introduce un número decimal: "

    Devuelve:
        El valor decimal introducido por el usuario (float).
    """
    while True:
        try:
            valor = float(input(mensaje))
        except ValueError:
            print("Entrada no válida. Debe ser un número decimal.")
            continue
        if math.isfinite(valor):                 # float() también acepta "inf" y "nan": no son valores útiles
            return valor
        print("Entrada no válida. No se admiten infinitos ni NaN.")


def crear_menu(opciones_menu: list[str]) -> int:
    """
    Función que muestra un menú de opciones y solicita al usuario
    que seleccione una opción válida.

    Parámetros:
        opciones_menu: lista de opciones a mostrar en el menú (list).

    Devuelve:
        El número de opción seleccionado por el usuario (int), empezando en 1.
    """
    if not opciones_menu:
        raise ValueError("El menú debe tener al menos una opción.")
    print()
    for numero, opcion in enumerate(opciones_menu, start=1):
        print(f"{numero}. {opcion}")
    while True:
        seleccion = leer_int("Seleccione una opción: ")
        if 1 <= seleccion <= len(opciones_menu):
            return seleccion
        print(f"Opción no válida. Elija un número entre 1 y {len(opciones_menu)}.")
````

````python
# matriz.py: clase MatrizFloat
# (en el archivo matriz.py aquí iría: from utils import leer_float)
from typing import Optional

import numpy as np


class MatrizFloat:
    """
    Clase que representa una matriz dinámica 1D/2D.

    Atributos de clase:
        TIPO_DATO: tipo de dato por defecto (np.float32)

    Atributos de instancia (protegidos):
        _matriz: almacena la matriz (numpy)
        _nfil: almacena número de filas de la matriz (int)
        _ncol: almacena número de columnas de la matriz (int)
    """

    TIPO_DATO: type = np.float32

    # Método de instancia: inicializa el estado propio de cada objeto (self).
    def __init__(self) -> None:
        """
        Inicializador. No recibe parámetros adicionales.
        Inicializa el atributo matriz con None y los atributos
        nfil y ncol a 0
        """
        self._matriz: Optional[np.ndarray] = None
        self._nfil: int = 0
        self._ncol: int = 0

    # Método de clase (@classmethod): es un constructor alternativo. No necesita un objeto
    # previo y usa cls para crear la instancia (y cls.TIPO_DATO para el tipo de dato).
    @classmethod
    def matriz_identidad(cls, n: int) -> "MatrizFloat":
        """
        Método de clase que crea y devuelve una nueva
        instancia de MatrizFloat configurada como matriz
        identidad de tamaño n x n

        Parámetros:
            n: el tamaño de la matriz (int, mayor que 0)

        Devuelve:
            Nueva instancia con matriz identidad creada (MatrizFloat)
        """
        cls._validar_dimension(n)
        nueva = cls()
        nueva._matriz = np.eye(n, dtype=cls.TIPO_DATO)      # eye: unos en la diagonal, ceros en el resto
        nueva._nfil = n
        nueva._ncol = n
        return nueva

    # Método estático (@staticmethod): recibe las dos matrices como parámetros y no usa
    # ningún atributo de una instancia concreta (self) ni de la clase (cls).
    @staticmethod
    def mismas_dimensiones(m1: "MatrizFloat", m2: "MatrizFloat") -> bool:
        """
        Método para comparar dimensiones.

        Parámetros:
            m1: primera matriz a comparar (MatrizFloat)
            m2: segunda matriz a comparar (MatrizFloat)

        Devuelve:
            bool: True si ambas matrices tienen el mismo número de
                  filas y columnas, y False en caso contrario
                  (o si alguna no está creada)
        """
        if not (m1.existe() and m2.existe()):
            return False
        return m1._nfil == m2._nfil and m1._ncol == m2._ncol

    # Método de instancia: crea y guarda la matriz en el objeto sobre el que se llama.
    def crear_matriz2D(self, n_fil: int, n_col: int) -> None:
        """
        Método para crear una matriz bidimensional de ceros.
        Asigna valores de filas y columnas según parámetros.

        Parámetros:
            n_fil: número de filas (int, mayor que 0)
            n_col: número de columnas (int, mayor que 0)
        """
        self._validar_dimension(n_fil)
        self._validar_dimension(n_col)
        self._matriz = np.zeros((n_fil, n_col), dtype=self.TIPO_DATO)
        self._nfil = n_fil
        self._ncol = n_col

    # Método de instancia: reutiliza crear_matriz2D sobre el mismo objeto.
    def crear_matriz1D(self, n_elementos: int) -> None:
        """
        Método para crear una matriz unidimensional de ceros.
        Usa crear_matriz2D para asignar 1 fila y n columnas.

        Parámetros:
            n_elementos: número de elementos de la fila (int, mayor que 0)
        """
        self.crear_matriz2D(1, n_elementos)

    # Método de instancia: consulta el estado de este objeto.
    def existe(self) -> bool:
        """
        Método que verifica si matriz está creada y no está vacía

        Devuelve:
            True si existe, de lo contrario False
        """
        return self._matriz is not None and self._matriz.size > 0

    # Método de instancia: rellena la matriz del propio objeto.
    def introducir(self) -> None:
        """
        Método que solicita por consola el valor numérico decimal
        para cada posición (i, j) de la matriz.

        Da error (ValueError) si se intenta introducir datos sin haber
        creado la matriz previamente.
        """
        if not self.existe():
            raise ValueError("No se pueden introducir datos: la matriz no ha sido creada.")
        for i in range(self._nfil):
            for j in range(self._ncol):
                self._matriz[i, j] = leer_float(f"Valor para la posición ({i}, {j}): ")

    # Método de instancia: muestra el estado del propio objeto.
    def mostrar(self) -> None:
        """
        Método para mostrar las dimensiones y los elementos de la
        matriz. Si la matriz no existe (existe() == False), muestra
        un mensaje advirtiéndolo
        """
        if not self.existe():
            print("La matriz no existe. Cree una matriz primero.")
            return
        print(f"Dimensiones: {self._nfil} x {self._ncol}")
        print(self._matriz)

    # ===========================================================
    # OPERACIONES ALGEBRAICAS
    # ===========================================================

    # Método de instancia: opera sobre la matriz actual (self) y otra matriz, sin modificar ninguna.
    def sumar_matrices(self, otra_matriz: "MatrizFloat") -> "MatrizFloat":
        """
        Método que suma la matriz actual con otra matriz.

        Parámetros:
            otra_matriz: objeto de MatrizFloat con la matriz a sumar

        Devuelve:
            Un nuevo objeto MatrizFloat con la matriz resultante de
            la suma.

        Lanza ValueError si las dimensiones de ambas matrices no coinciden.
        """
        self._comprobar_dimensiones(otra_matriz)
        resultado = MatrizFloat()
        resultado.crear_matriz2D(self._nfil, self._ncol)
        resultado._matriz[:] = self._matriz + otra_matriz._matriz
        return resultado

    # Método de instancia: igual que sumar_matrices, pero restando.
    def restar_matrices(self, otra_matriz: "MatrizFloat") -> "MatrizFloat":
        """
        Método que resta la matriz actual con otra matriz.

        Parámetros:
            otra_matriz: objeto de MatrizFloat con la matriz a restar

        Devuelve:
            Un nuevo objeto MatrizFloat con la matriz resultante de
            la resta.

        Lanza ValueError si las dimensiones de ambas matrices no coinciden.
        """
        self._comprobar_dimensiones(otra_matriz)
        resultado = MatrizFloat()
        resultado.crear_matriz2D(self._nfil, self._ncol)
        resultado._matriz[:] = self._matriz - otra_matriz._matriz
        return resultado

    # Método estático auxiliar (protegido): valida un número, no necesita self ni cls.
    @staticmethod
    def _validar_dimension(valor: int) -> None:
        """Lanza ValueError si valor no es un entero mayor que 0."""
        if not isinstance(valor, (int, np.integer)) or isinstance(valor, bool) or valor < 1:
            raise ValueError(f"Las dimensiones deben ser enteros mayores que 0 (recibido: {valor!r}).")

    # Método de instancia auxiliar (protegido): comprueba que se puede operar con otra matriz.
    def _comprobar_dimensiones(self, otra_matriz: "MatrizFloat") -> None:
        """Lanza ValueError si alguna matriz no existe o si sus dimensiones no coinciden."""
        if not (self.existe() and otra_matriz.existe()):
            raise ValueError("Ambas matrices deben estar creadas para operar con ellas.")
        if not MatrizFloat.mismas_dimensiones(self, otra_matriz):
            raise ValueError(
                f"Las dimensiones no coinciden: {self._nfil}x{self._ncol} y "
                f"{otra_matriz._nfil}x{otra_matriz._ncol}."
            )
````

````python
# Programa principal
# (en un archivo aparte iría: from matriz import MatrizFloat y from utils import leer_int, crear_menu)
OPCIONES_PRINCIPAL = [
    "Construir matriz 1D",
    "Construir matriz 2D",
    "Construir matriz identidad",
    "Introducir matriz",
    "Mostrar matriz",
    "Operaciones con matrices",
    "Terminar",
]

OPCIONES_OPERACIONES = [
    "Sumar matrices",
    "Restar matrices",
    "Volver al menú principal",
]


def operar_con_otra_matriz(matriz: MatrizFloat, es_suma: bool) -> None:
    """
    Pide al usuario una segunda matriz, la suma o resta a 'matriz' y muestra el resultado.

    Parámetros:
        matriz: primera matriz de la operación (MatrizFloat)
        es_suma: True para sumar, False para restar (bool)
    """
    if not matriz.existe():
        print("Error: primero debe crear la matriz principal (opciones 1, 2 o 3).")
        return
    print("Datos de la segunda matriz:")
    segunda = MatrizFloat()
    segunda.crear_matriz2D(leer_int("Número de filas: "), leer_int("Número de columnas: "))
    if not MatrizFloat.mismas_dimensiones(matriz, segunda):     # se avisa antes de pedir todos los valores
        print("Error: las dimensiones de ambas matrices deben coincidir.")
        return
    segunda.introducir()
    resultado = matriz.sumar_matrices(segunda) if es_suma else matriz.restar_matrices(segunda)
    print("Resultado de la suma:" if es_suma else "Resultado de la resta:")
    resultado.mostrar()


def submenu_operaciones(matriz: MatrizFloat) -> None:
    """Muestra el submenú de operaciones hasta que el usuario elija volver al menú principal."""
    while True:
        opcion = crear_menu(OPCIONES_OPERACIONES)
        if opcion == 3:
            return
        try:
            operar_con_otra_matriz(matriz, es_suma=(opcion == 1))
        except ValueError as error:
            print(f"Error: {error}")


def main() -> None:
    """Ejecuta el menú principal hasta que el usuario elija terminar."""
    matriz = MatrizFloat()
    while True:
        opcion = crear_menu(OPCIONES_PRINCIPAL)
        try:
            if opcion == 1:
                matriz = MatrizFloat()
                matriz.crear_matriz1D(leer_int("Número de elementos: "))
                print("Matriz 1D creada.")
            elif opcion == 2:
                matriz = MatrizFloat()
                matriz.crear_matriz2D(leer_int("Número de filas: "), leer_int("Número de columnas: "))
                print("Matriz 2D creada.")
            elif opcion == 3:
                matriz = MatrizFloat.matriz_identidad(leer_int("Tamaño n de la matriz identidad: "))
                print("Matriz identidad creada.")
            elif opcion == 4:
                matriz.introducir()
            elif opcion == 5:
                matriz.mostrar()
            elif opcion == 6:
                submenu_operaciones(matriz)
            else:
                print("Hasta pronto.")
                break
        except ValueError as error:
            print(f"Error: {error}")
````
